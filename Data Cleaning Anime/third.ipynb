```python
anime = anime[['Rank', 'Name', 'Tags', 'Rating']]
```

```python
anime.shape
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features= 5000, stop_words='english')
```

```python
vectors = cv.fit_transform(anime['Tags']).toarray()
```

```python
count=0
number = 0
for i in cv.get_feature_names_out():
    number+=1
    if (i == 'movie'):
        count +=1
        break
print(count, number)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    
    return " ".join(y)
```

```python
anime['Tags'] = anime['Tags'].apply(stem)
```

```python
vectors.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
anime[anime['Name'] == 'Naruto'].index[0]
```

```python
(sorted(list(enumerate(similarity[0])),reverse=True, key = lambda x:x[1])[1:6])
```

```python
def recommend(anime_name):
    anime_index = anime[anime['Name'] == anime_name].index[0]
    distances = similarity[anime_index] #array
    anime_list = sorted(list(enumerate(distances)),reverse=True, key = lambda x:x[1])[1:11] #gives anime index along with distance
    
    for i in anime_list:
        print(anime.iloc[i[0]]['Name'])
```

```python
recommend('In Another World With My Smartphone')
```
